```python
# Constants

# Training data
TRAIN_DATA = "C:\\Users\\caleb\\OneDrive - Embry-Riddle Aeronautical University\\Documents\\Coding\\Pytorch\\Pytorch\\Data\\frankdatayolov1.1\\data.yaml"

# Project output directory
PROJECT_OUTPUT = "C:/Users/caleb/OneDrive - Embry-Riddle Aeronautical University/Documents/Coding/Pytorch/Models/Model Files/frank1"

MODEL_SELECTION = "yolov8n.pt"  # Switch the n with s, m, l, x for different model sizes

OUTPUT_NAME = "frank1results"   # Name for the output results folder

OPTIMIZER_SELECTION = "Adam"  # Options: 'SGD', 'Adam', 'AdamW'

NUM_EPOCHS = 100 # How many times to loop through the dataset

IMAGE_SIZE = 1920 # Resolution of input images

BATCH_SIZE = 4 # Number of images per batch

DEVICE = 0  # device = mps for apple silicon, 0 for first GPU, -1 for CPU

SEED = 42 # Random seed for reproducibility

DETERMINISTIC = True # Whether to use deterministic training for reproducibility

LR0 = 1E-3 # Initial learning rate

ROTATE_DEGREES = .25 # Degree of rotation for data augmentation

TRANSLATE = .1 # Translation factor for data augmentation

SCALE = .5 # Scaling factor for data augmentation

SHEAR = 90 # Shear angle for data augmentation

PERSPECTIVE = 0.0005 # Perspective factor for data augmentation

FLIPLR = .2 # Left-right flip probability for data augmentation

MOSAIC = .25 # Mosaic augmentation probability

MIXUP = .15 # Mixup augmentation probability

CUTMIX = .5 # CutMix augmentation probability

COPY_PASTE = .5 # Copy-paste augmentation probability

ERASING = .2 # Random erasing probability
```

```python
# Imports
import torch
from ultralytics import YOLO
```

```python
print(f"PyTorch version: {torch.__version__}")
print(f"CUDA available: {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"CUDA device: {torch.cuda.get_device_name(0)}")
    print(f"CUDA version: {torch.version.cuda}")
    print(f"GPU Memory Usage: {torch.cuda.memory_allocated()/1024**2:.2f} MB / {torch.cuda.max_memory_allocated()/1024**2:.2f} MB")

```

```python
# Load a pretrained yolo model

# For YOLOv8 nano model configuration
model = YOLO(MODEL_SELECTION)
```

```python
import torch
torch.cuda.empty_cache()
```

```python
# Training
results = model.train(
    data=TRAIN_DATA,
    epochs=NUM_EPOCHS, 
    imgsz=IMAGE_SIZE, 
    batch=BATCH_SIZE,
    device=DEVICE, #device = mps for apple silicon
    save_period=1,
    project=PROJECT_OUTPUT,
    name=OUTPUT_NAME,
    optimizer=OPTIMIZER_SELECTION,
    seed=SEED,
    deterministic=DETERMINISTIC,
    lr0=LR0,
    plots=True,
    degrees=ROTATE_DEGREES,
    translate=TRANSLATE,
    scale=SCALE,
    shear=SHEAR,
    perspective=PERSPECTIVE,
    fliplr=FLIPLR,
    mosaic=MOSAIC,
    mixup=MIXUP,
    cutmix=CUTMIX,
    copy_paste=COPY_PASTE,
    erasing=ERASING,
)
```

```python
# Validation

results = model.val(data = "C:\\Users\\caleb\\OneDrive - Embry-Riddle Aeronautical University\\Documents\\Coding\\Pytorch\\Pytorch\\Data\\frankdatayolov1.1\\data.yaml")
```
