```python
# Constants

#-------------------------- For Training ------------------------ #

# Training data
TRAIN_DATA = "C:\\Users\\caleb\\OneDrive - Embry-Riddle Aeronautical University\\Documents\\Coding\\Pytorch\\Pytorch\\Data\\frankdatayolov1.1\\data.yaml"

# Project output directory
PROJECT_OUTPUT = "C:/Users/caleb/OneDrive - Embry-Riddle Aeronautical University/Documents/Coding/Pytorch/Models/Model Files/frank1"

MODEL_SELECTION = "yolov8n.pt"  # Switch the n with s, m, l, x for different model sizes

OUTPUT_NAME = "frank1results"   # Name for the output results folder

OPTIMIZER_SELECTION = 'auto' # Options: 'SGD', 'Adam', 'AdamW', 'RMSprop', or 'auto' for automatic selection

NUM_EPOCHS = 100 # How many times to loop through the dataset

IMAGE_SIZE = 1920 # Resolution of input images

BATCH_SIZE = 4 # Number of images per batch

DEVICE = 0  # device = mps for apple silicon, 0 for first GPU, -1 for CPU

SEED = 42 # Random seed for reproducibility

DETERMINISTIC = True # Whether to use deterministic training for reproducibility

LR0 = 1E-3 # Initial learning rate

ROTATE_DEGREES = .25 # Degree of rotation for data augmentation

TRANSLATE = .1 # Translation factor for data augmentation

SCALE = .5 # Scaling factor for data augmentation

SHEAR = 90 # Shear angle for data augmentation

PERSPECTIVE = 0.0005 # Perspective factor for data augmentation

FLIPLR = .2 # Left-right flip probability for data augmentation

MOSAIC = .25 # Mosaic augmentation probability

MIXUP = .15 # Mixup augmentation probability

CUTMIX = .5 # CutMix augmentation probability

COPY_PASTE = .5 # Copy-paste augmentation probability

ERASING = .2 # Random erasing probability

SAVE_PERIOD = -1 # Save model every n epochs or -1 to disable

MAX_TRAIN_TIME = None # Maximum training time in seconds or None for no limit !!! Overides NUM_EPOCHS if set !!!

SINGLE_CLASS = True # Whether to treat the dataset as single-class

COS_LR = True # Whether to use cosine learning rate schedule

WARMUP_EPOCHS = 3.0 # Number of warmup epochs

COMPILE_MODEL = True # Whether to compile the model for faster training



#------------------------ For Validation ------------------------- #

IOU = 0.5 # Intersection over Union threshold for validation

COMPILE = True # Whether to compile the model for faster inference

PLOTTING = True # Whether to generate plots during validation

VAL_PLOTS = True # Whether to generate additional plots during validation

VERBOSE = True # Whether to print detailed validation results

HALF = False # Whether to use half precision during validation


#-------------------------- For Inference ------------------------ #

CONFIDENCE_THRESHOLD = 0.5 # Confidence threshold for predictions
```

```python
# Imports
import torch
from ultralytics import YOLO
```

```python
print(f"PyTorch version: {torch.__version__}")
print(f"CUDA available: {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"CUDA device: {torch.cuda.get_device_name(0)}")
    print(f"CUDA version: {torch.version.cuda}")
    print(f"GPU Memory Usage: {torch.cuda.memory_allocated()/1024**2:.2f} MB / {torch.cuda.max_memory_allocated()/1024**2:.2f} MB")

```

```python
# Load a pretrained yolo model

# For YOLOv8 nano model configuration
model = YOLO(MODEL_SELECTION)
```

```python
import torch
torch.cuda.empty_cache()
```

```python
# Training
results = model.train(
    data=TRAIN_DATA,
    epochs=NUM_EPOCHS, 
    imgsz=IMAGE_SIZE, 
    batch=BATCH_SIZE,
    device=DEVICE,
    save_period=SAVE_PERIOD,
    project=PROJECT_OUTPUT,
    name=OUTPUT_NAME,
    optimizer=OPTIMIZER_SELECTION,
    seed=SEED,
    deterministic=DETERMINISTIC,
    lr0=LR0,
    plots=True,
    degrees=ROTATE_DEGREES,
    translate=TRANSLATE,
    scale=SCALE,
    shear=SHEAR,
    perspective=PERSPECTIVE,
    fliplr=FLIPLR,
    mosaic=MOSAIC,
    mixup=MIXUP,
    cutmix=CUTMIX,
    copy_paste=COPY_PASTE,
    erasing=ERASING,
    time = MAX_TRAIN_TIME,
    single_cls = SINGLE_CLASS,
    cos_lr = COS_LR,
    warmup_epochs = WARMUP_EPOCHS,
    compile=COMPILE_MODEL,
)
```

```python
# Validation

results = model.val(
    data = "C:\\Users\\caleb\\OneDrive - Embry-Riddle Aeronautical University\\Documents\\Coding\\Pytorch\\Pytorch\\Data\\frankdatayolov1.1\\data.yaml",
    iou=IOU,
    device=DEVICE,
    plots=VAL_PLOTS,
    half=HALF,
    verbose=VERBOSE,
    compile=COMPILE,
)   
```

```python
# Export

success = model.export(format="onnx", dir="C:\Users\caleb\OneDrive - Embry-Riddle Aeronautical University\Documents\Coding\Pytorch\Models\Model Files\frank1")  # export the model to ONNX format
minimodel= model.export(format="ncnn", dir="C:\Users\caleb\OneDrive - Embry-Riddle Aeronautical University\Documents\Coding\Pytorch\Models\Model Files\frank1") # export the model to NCNN format
```
